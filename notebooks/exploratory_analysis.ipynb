# Exploratory analysis
Run `scripts/prepare_data.py` first. This notebook only reads files that exist; it draws nothing from invented data.

```python
import json, pathlib
import matplotlib.pyplot as plt
ROOT = pathlib.Path('..').resolve()
stats = json.loads((ROOT/'data/processed/stats.json').read_text())
meta = json.loads((ROOT/'data/processed/metadata.json').read_text())
print(meta['dataset'])
```

```python
for split, s in stats['splits'].items():
    print(split, s['num_images'], 'images', s['num_captions'], 'captions', 'mean len', round(s.get('caption_len_mean', 0), 2), 'OOV', round(s.get('oov_rate', 0), 4))
```

```python
hist = stats['splits']['train']['caption_len_hist']
plt.bar(range(len(hist)), hist)
plt.xlabel('caption length (tokens; 40 = 40+)'); plt.ylabel('captions'); plt.title('Training caption lengths'); plt.show()
```

```python
import random
recs = [r for r in meta['records'] if r['split'] == 'train']
for r in random.Random(0).sample(recs, 3):
    print(r['image_id'], r['file_name'])
    for c in r['captions']: print('  -', c)
```
